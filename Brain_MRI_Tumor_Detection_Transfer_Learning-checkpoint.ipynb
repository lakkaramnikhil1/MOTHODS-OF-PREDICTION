## Brain MRI Images for Brain Tumor Detection

## Introduction

Brain tumors are an important medical-imaging problem because distinguishing abnormal brain tissue may aid in additional clinical evaluation and planning of treatment. When brain abnormalities are under consideration, they often will use Magnetic Resonance Imaging (MRI) to get detailed pictures of the structures of the brain.

Deep learning, specifically Convolutional Neural Networks (CNNs), has the ability to learn visual patterns from images directly. New CNN-based systems do not need to be manually designed to learn spatial, textural, and structural representations from MRI scans.

In this project, end-to-end deep-learning pipeline that automatically detects binary brain tumor from MRI image is built. It features data exploration, data quality assessment, duplicate detection, data preprocessing, stratified splitting, transfer learning with MobileNetV2 model, 10 different experimental pipeline setups, model selection, model fine tuning, validation based threshold selection, and model explanation.


## Problem Statement

This project aims at finding out if there's any sign of a tumor in an MRI of the brain. The problem is specified as an image classification problem in a supervised manner and with two classes that the problem solves:

- **Class 0 — No Tumor**
- **Class 1 — Tumor**

As a healthcare-technology aspect, this kind of model might help to screen or prioritise indeterminate MRI scans for different MRI specialists to review. It will not supplant radiologists or serve as an independent medical diagnosis; rather, it will be considered as a proof-of-concept decision support model.

The database has 253 original MRI images with tumor and non-tumour classes. This data set is too small to train a large CNN from scratch, instead, it uses transfer learning, which uses features extracted from an extremely large image data set to help achieve the large CNN's vision.

## Import Libraries

```python
import zipfile
import hashlib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from PIL import Image
from IPython.display import display, Markdown
from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay, roc_auc_score, RocCurveDisplay, roc_curve
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

tf.keras.utils.set_random_seed(42)
```

## Unzip Dataset

```python
archive_path = Path("archive.zip")
extract_dir = Path("brain_mri_extracted")
valid_ext = {".jpg", ".jpeg", ".png"}

if archive_path.exists():
    extract_dir.mkdir(exist_ok=True)
    with zipfile.ZipFile(archive_path, "r") as z:
        z.extractall(extract_dir)

search_root = extract_dir if extract_dir.exists() else Path(".")
candidates = []

for p in [search_root] + [x for x in search_root.rglob("*") if x.is_dir()]:
    yes_dir = p / "yes"
    no_dir = p / "no"
    if yes_dir.is_dir() and no_dir.is_dir():
        total = sum(1 for f in yes_dir.iterdir() if f.suffix.lower() in valid_ext)
        total += sum(1 for f in no_dir.iterdir() if f.suffix.lower() in valid_ext)
        candidates.append((total, p))

if not candidates:
    raise FileNotFoundError("No folder containing both 'yes' and 'no' was found.")

data_dir = sorted(candidates, key=lambda x: x[0], reverse=True)[0][1]
print("Dataset directory:", data_dir)
```

## Dataset Indexing

```python
records = []

for class_name, label in [("no", 0), ("yes", 1)]:
    for p in sorted((data_dir / class_name).iterdir()):
        if p.suffix.lower() in valid_ext:
            records.append({
                "filepath": str(p),
                "class_name": class_name,
                "label": label
            })

df = pd.DataFrame(records)
display(df.head())
print("Total images:", len(df))
display(df["class_name"].value_counts().to_frame("count"))
```

Mean pixel intensity is a straightforward measure of an image's brightness. The mean grayscale intensity values of tumor vs non-tumor (NT) MRI images are examined and compared in the next step. ### Duplicate Check

```python
def sha256_file(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        h.update(f.read())
    return h.hexdigest()

df = df[df["readable"]].copy()
df["hash"] = df["filepath"].apply(sha256_file)

duplicate_count = int(df.duplicated("hash").sum())
conflict_hashes = df.groupby("hash")["label"].nunique()
conflict_hashes = conflict_hashes[conflict_hashes > 1].index

print("Exact duplicates:", duplicate_count)
print("Conflicting duplicate hashes:", len(conflict_hashes))

df = df[~df["hash"].isin(conflict_hashes)].drop_duplicates("hash").reset_index(drop=True)
print("Images retained:", len(df))
```

## Train Validation Test Split

```python
train_df, temp_df = train_test_split(
    df,
    test_size=0.30,
    stratify=df["label"],
    random_state=42
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    stratify=temp_df["label"],
    random_state=42
)

print("Training:", len(train_df))
print("Validation:", len(val_df))
print("Test:", len(test_df))

display(pd.DataFrame({
    "train": train_df["class_name"].value_counts(),
    "validation": val_df["class_name"].value_counts(),
    "test": test_df["class_name"].value_counts()
}).fillna(0).astype(int))
```

## Data Preprocessing

```python
IMG_SIZE = (224, 224)
AUTOTUNE = tf.data.AUTOTUNE

def load_image(path, label):
    image = tf.io.read_file(path)
    image = tf.io.decode_image(image, channels=3, expand_animations=False)
    image.set_shape([None, None, 3])
    image = tf.image.resize(image, IMG_SIZE)
    image = tf.cast(image, tf.float32)
    return image, tf.cast(label, tf.float32)

def make_dataset(frame, batch_size=16, training=False):
    paths = frame["filepath"].astype(str).values
    labels = frame["label"].astype(np.float32).values
    ds = tf.data.Dataset.from_tensor_slices((paths, labels))
    if training:
        ds = ds.shuffle(len(frame), seed=42, reshuffle_each_iteration=True)
    ds = ds.map(load_image, num_parallel_calls=AUTOTUNE)
    ds = ds.batch(batch_size).prefetch(AUTOTUNE)
    return ds

train_ds = make_dataset(train_df, training=True)
val_ds = make_dataset(val_df)
test_ds = make_dataset(test_df)
```

## CNN and Transfer Learning Model Architecture

### Why a Convolutional Neural Network?

CNNs are developed to handle image data as they contain local spatial patterns and maintain the relationship among neighbouring pixels. Earlier layers are more likely to extract lower level features like edges, intensity transitions and textures, and deeper layers will compose them into higher level features of images.

The data set is small after removal of duplicate data, and this data is unfit for deep learning. Starting a big CNN from random initialisations would then run a significant risk of overfitting. Consequently, as part of this project, the use of the MobileNetV2 feature extractor, pre-trained on ImageNet data, is relied on.

The modeled changes were developed using a layer-by-layer approach.A layer-by-layer explanation of the changes modeled was used.

### Layer-by-Layer Explanation of the Implemented Model

#### 1. Input Layer

The model receives an MRI tensor with shape **224 × 224 × 3**. The first two dimensions are image height and width, while the final dimension contains the three image channels.

#### 2. Data Augmentation

For experimental configurations where augmentation is enabled, the model can apply:

- `RandomFlip("horizontal")`
- `RandomRotation(...)`
- `RandomZoom(...)`

These transformations increase training-data variation and are intended to reduce overfitting. Their strength is changed across the experimental configurations.

#### 3. MobileNetV2 Preprocessing

`tf.keras.applications.mobilenet_v2.preprocess_input()` transforms the image values into the numerical representation expected by the pretrained MobileNetV2 network.

#### 4. MobileNetV2 Convolutional Backbone

MobileNetV2 has been loaded with include_top=False, discarding its built-in image classification model ImageNet and keeping the convolutional feature extractor. First the backbone is frozen (“backbone.trainable” is set to False) to ensure that the pretrained convolutional weights are kept in the first training phase.

The backbone generates high-level features maps having learned visual information for a 224 × 224 input. MobileNetV2 architecture adopts the efficient depthwise-separable convolution and inverted-residual structures for transfer learning without the risk of excessive parameters.

#### 5. Global Average Pooling

`GlobalAveragePooling2D()` converts the spatial feature maps produced by MobileNetV2 into a one-dimensional feature vector by averaging each feature channel. This greatly reduces the number of parameters compared with flattening the entire feature map and helps reduce overfitting.

#### 6. First Dropout Layer

The first `Dropout` layer randomly disables a proportion of feature activations during training. Different dropout values are evaluated in the experiments, including 0.20, 0.30, 0.40 and 0.50. This acts as regularisation and reduces dependence on individual activations.

#### 7. Dense Hidden Layer

A fully connected `Dense` layer combines the extracted MobileNetV2 features. It uses **ReLU** activation to introduce non-linearity. The experiments evaluate different hidden-layer sizes, including 64, 128 and 256 units, so the effect of classification-head capacity can be compared.

#### 8. Second Dropout Layer

A second `Dropout` layer is placed after the Dense layer to provide additional regularisation before the final classification output.

#### 9. Sigmoid Output Layer

The final `Dense(1, activation="sigmoid")` layer contains one output neuron. Its value lies between 0 and 1 and represents the model's estimated probability of the **tumor** class. A later validation-selected decision threshold converts this probability into the final binary prediction.

### Loss Function

The model uses **binary cross-entropy** because the task contains two classes and the output layer uses a sigmoid probability.

### Optimiser

The **Adam optimiser** updates the trainable parameters. Different learning rates are tested because the learning rate can substantially affect optimisation and generalisation.

### Evaluation Metrics

Training and validation monitor:

- **Accuracy** — overall proportion of correct predictions.
- **Precision** — proportion of predicted tumor cases that are actually tumor cases.
- **Recall** — proportion of actual tumor cases correctly detected.
- **ROC-AUC** — ability of predicted probabilities to discriminate between the two classes across possible decision thresholds.

Multiple metrics are important as accurate ranking does not convey the leeway for false positive versus false negative predictions.


## Transfer Learning Model Builder

```python
def build_transfer_model(config):
    augmentation = keras.Sequential([
        layers.RandomFlip("horizontal"),
        layers.RandomRotation(config["rotation"]),
        layers.RandomZoom(config["zoom"])
    ], name="augmentation")

    backbone = tf.keras.applications.MobileNetV2(
        include_top=False,
        weights="imagenet",
        input_shape=IMG_SIZE + (3,)
    )

    backbone.trainable = False

    inputs = keras.Input(shape=IMG_SIZE + (3,))
    x = augmentation(inputs) if config["augmentation"] else inputs
    x = tf.keras.applications.mobilenet_v2.preprocess_input(x)
    x = backbone(x, training=False)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dropout(config["dropout"])(x)
    x = layers.Dense(config["dense_units"], activation="relu")(x)
    x = layers.Dropout(config["head_dropout"])(x)
    outputs = layers.Dense(1, activation="sigmoid")(x)

    model = keras.Model(inputs, outputs)

    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=config["learning_rate"]),
        loss="binary_crossentropy",
        metrics=[
            keras.metrics.BinaryAccuracy(name="accuracy"),
            keras.metrics.Precision(name="precision"),
            keras.metrics.Recall(name="recall"),
            keras.metrics.AUC(name="auc")
        ]
    )

    return model
```

## Experimental Design

### Ten Pipeline Configurations

```python
experiments = [
    {"name": "TL Baseline", "dropout": 0.20, "head_dropout": 0.10, "dense_units": 64, "learning_rate": 0.0010, "augmentation": False, "rotation": 0.00, "zoom": 0.00, "class_weight": False},
    {"name": "TL Dropout 0.3", "dropout": 0.30, "head_dropout": 0.20, "dense_units": 64, "learning_rate": 0.0010, "augmentation": False, "rotation": 0.00, "zoom": 0.00, "class_weight": False},
    {"name": "TL Dropout 0.5", "dropout": 0.50, "head_dropout": 0.30, "dense_units": 64, "learning_rate": 0.0010, "augmentation": False, "rotation": 0.00, "zoom": 0.00, "class_weight": False},
    {"name": "TL Dense 128", "dropout": 0.30, "head_dropout": 0.20, "dense_units": 128, "learning_rate": 0.0010, "augmentation": False, "rotation": 0.00, "zoom": 0.00, "class_weight": False},
    {"name": "TL Dense 256", "dropout": 0.30, "head_dropout": 0.20, "dense_units": 256, "learning_rate": 0.0010, "augmentation": False, "rotation": 0.00, "zoom": 0.00, "class_weight": False},
    {"name": "TL Lower LR", "dropout": 0.30, "head_dropout": 0.20, "dense_units": 128, "learning_rate": 0.0003, "augmentation": False, "rotation": 0.00, "zoom": 0.00, "class_weight": False},
    {"name": "TL Augmentation", "dropout": 0.30, "head_dropout": 0.20, "dense_units": 128, "learning_rate": 0.0003, "augmentation": True, "rotation": 0.05, "zoom": 0.10, "class_weight": False},
    {"name": "TL Stronger Augmentation", "dropout": 0.30, "head_dropout": 0.20, "dense_units": 128, "learning_rate": 0.0003, "augmentation": True, "rotation": 0.08, "zoom": 0.15, "class_weight": False},
    {"name": "TL Class Weighted", "dropout": 0.30, "head_dropout": 0.20, "dense_units": 128, "learning_rate": 0.0003, "augmentation": True, "rotation": 0.05, "zoom": 0.10, "class_weight": True},
    {"name": "TL Low LR Weighted", "dropout": 0.40, "head_dropout": 0.20, "dense_units": 128, "learning_rate": 0.0001, "augmentation": True, "rotation": 0.05, "zoom": 0.10, "class_weight": True}
]

display(pd.DataFrame(experiments))
```

### Run Ten Experiments

```python
results = []

for experiment_id, config in enumerate(experiments, start=1):
    tf.keras.backend.clear_session()
    tf.keras.utils.set_random_seed(42)

    model = build_transfer_model(config)

    callbacks = [
        keras.callbacks.EarlyStopping(
            monitor="val_auc",
            mode="max",
            patience=4,
            restore_best_weights=True
        ),
        keras.callbacks.ReduceLROnPlateau(
            monitor="val_loss",
            factor=0.3,
            patience=2,
            min_lr=1e-6
        )
    ]

    history = model.fit(
        train_ds,
        validation_data=val_ds,
        epochs=20,
        verbose=0,
        class_weight=class_weights if config["class_weight"] else None,
        callbacks=callbacks
    )

    scores = model.evaluate(val_ds, verbose=0, return_dict=True)
    best_epoch = int(np.argmax(history.history["val_auc"]) + 1)

    results.append({
        "experiment": experiment_id,
        "configuration": config["name"],
        "val_accuracy": scores["accuracy"],
        "val_precision": scores["precision"],
        "val_recall": scores["recall"],
        "val_auc": scores["auc"],
        "best_epoch": best_epoch,
        "epochs_run": len(history.history["loss"])
    })

results_df = pd.DataFrame(results).sort_values(
    "val_auc",
    ascending=False
).reset_index(drop=True)

display(results_df)
```

## Best Configuration Selection

```python
best_row = results_df.iloc[0]
best_experiment = int(best_row["experiment"])
best_config = experiments[best_experiment - 1]

print("Selected experiment:", best_experiment)
print("Selected configuration:", best_config["name"])
display(best_row.to_frame().T)
```

## Final Transfer Model Training

```python
tf.keras.backend.clear_session()
tf.keras.utils.set_random_seed(42)

best_model = build_transfer_model(best_config)

callbacks = [
    keras.callbacks.EarlyStopping(
        monitor="val_auc",
        mode="max",
        patience=5,
        restore_best_weights=True
    ),
    keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.3,
        patience=2,
        min_lr=1e-6
    )
]

initial_history = best_model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=30,
    verbose=1,
    class_weight=class_weights if best_config["class_weight"] else None,
    callbacks=callbacks
)
```

## Fine Tuning

### Fine-Tuning Strategy

Once trained, the trained classification head is unfrozen from the notebook while all the other layers are frozen. Batch-normalisation layers are also kept frozen and the model is rebuilt with a smaller learning rate (`1e-5`).

The idea behind fine tuning is to enable the higher-level pre-trained features' adaptation to the specific task of brain MRI while avoiding significant modifications to the entire pre-trained network. It should be validated carefully with the original frozen-backbone configuration, not assumed that the tuning is responsible for a better performance than the original configuration.

```python
backbone = next(
    layer for layer in best_model.layers
    if isinstance(layer, keras.Model) and "mobilenetv2" in layer.name.lower()
)

backbone.trainable = True

for layer in backbone.layers[:-30]:
    layer.trainable = False

for layer in backbone.layers:
    if isinstance(layer, layers.BatchNormalization):
        layer.trainable = False

best_model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-5),
    loss="binary_crossentropy",
    metrics=[
        keras.metrics.BinaryAccuracy(name="accuracy"),
        keras.metrics.Precision(name="precision"),
        keras.metrics.Recall(name="recall"),
        keras.metrics.AUC(name="auc")
    ]
)

fine_callbacks = [
    keras.callbacks.EarlyStopping(
        monitor="val_auc",
        mode="max",
        patience=5,
        restore_best_weights=True
    ),
    keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.3,
        patience=2,
        min_lr=1e-7
    )
]

fine_history = best_model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=20,
    verbose=1,
    class_weight=class_weights if best_config["class_weight"] else None,
    callbacks=fine_callbacks
)

best_model.summary()
```

## Validation Threshold Selection

```python
val_probabilities = best_model.predict(val_ds, verbose=0).ravel()
val_labels = val_df["label"].values

fpr_val, tpr_val, thresholds_val = roc_curve(
    val_labels,
    val_probabilities
)

youden_j = tpr_val - fpr_val
finite_thresholds = np.isfinite(thresholds_val)
threshold_scores = np.where(finite_thresholds, youden_j, -np.inf)
best_threshold_index = int(np.argmax(threshold_scores))
best_threshold = float(thresholds_val[best_threshold_index])

val_predictions = (val_probabilities >= best_threshold).astype(int)
val_report = classification_report(
    val_labels,
    val_predictions,
    target_names=["no", "yes"],
    output_dict=True,
    zero_division=0
)

validation_table = pd.DataFrame({
    "Metric": [
        "Threshold",
        "Accuracy",
        "Precision",
        "Recall",
        "F1-score",
        "ROC-AUC"
    ],
    "Value": [
        best_threshold,
        float((val_predictions == val_labels).mean()),
        val_report["yes"]["precision"],
        val_report["yes"]["recall"],
        val_report["yes"]["f1-score"],
        roc_auc_score(val_labels, val_probabilities)
    ]
})

display(validation_table)
```

## Model Assessment on Unseen Test Set

```python
test_probabilities = best_model.predict(test_ds, verbose=0).ravel()
test_labels = test_df["label"].values
test_predictions = (test_probabilities >= best_threshold).astype(int)

report = classification_report(
    test_labels,
    test_predictions,
    target_names=["no", "yes"],
    output_dict=True,
    zero_division=0
)

test_accuracy = float((test_predictions == test_labels).mean())
test_auc = roc_auc_score(test_labels, test_probabilities)
test_f1 = report["yes"]["f1-score"]

metrics_table = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1-score",
        "ROC-AUC",
        "Decision Threshold"
    ],
    "Value": [
        test_accuracy,
        report["yes"]["precision"],
        report["yes"]["recall"],
        test_f1,
        test_auc,
        best_threshold
    ]
})

display(metrics_table)
```

## Discussion

The MobileNetV2 transfer-learning pipeline was successful under the 10 experimental setups, whereas tests indicated that the dropout, learning rate, Dense-layer size, augmentation and class weighting settings had an impact on the validation performance.

The best original configuration was **TL Dropout 0.5** which obtained the validation ROC-AUC of around **0.9853**. Even on the baselines model it worked well, and there was no training improvement with Dense-layer size. The higher the augmentation, the more accuracy and sensitivity were reduced but the higher the degree of tumor recall, indicating that there was some compromise between tumor recall and both accuracy and sensitivity.

Fine-tuning did not improve the best validation result, as ROC-AUC decreased from approximately **0.9853** to **0.9780**.

On the unseen test set, the final model achieved **0.829 accuracy, 0.864 precision, 0.864 recall, 0.864 F1-score and 0.864 ROC-AUC**. There are a total of 19 true positives, 10 true negatives, 3 false positives and 3 false negatives in the confusion matrix.

Overall, the model was able to provide useful classification ability and even though its test performance was poorer than the validation performance, it was limiting generalisation, primarily due to the size of the dataset.

## Limitations

The major drawback is the fact that the data set was small. Following the removal of 25 exact duplicates, there were 228 unique images of MRI scans, with 159 of these being used for training, 34 for validation, and 35 for testing.

The data is also somewhat unbalanced as there are 141 tumor images and 87 non-tumour images. Also, the data could not be measured in different hospitals, MRI scanners, or patients.

The model can only make a classification between 'tumor' or 'no-tumor' and cannot report the tumor type, the tumor grade, tumor size or location.

The drop in performance when the test set is different from the validation set is also poor generalisation. Lastly, the saliency map is only partially explainable and cannot be used to confirm that highlighted areas are actually tumour tissue.

## Recommendations

For future research, a much larger and diversified MRI-based collection done at multiple hospitals and with multiple types of scanners.

Avoiding data leakage, use patient level splitting if the patient has more than one image.

ResNet, DenseNet, and EfficientNet are other transfer-learning architectures that could be compared with the aforementioned architecture.

Grad-CAM combined with expertise from medical doctors may enhance the explainability.

There is no intention of using the current model as an independent diagnostic tool. It should be considered a proof-of-concept and used with the knowledge that it needs to be expanded with greater sizes of datasets and confirmed in an external clinical setting.
